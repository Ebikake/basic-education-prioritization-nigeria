# 02 BENI Index Construction

This notebook prepares the final BENI score files for the capstone project:

**A Data-Driven Framework for Prioritizing Public Capital Allocation to Basic Education Infrastructure across Nigerian States Using Education Need Indicators**

## Purpose

This notebook documents and exports:

1. BENI-Core model-ready dataset.
2. BENI-Expanded 2022 model-ready dataset.
3. BENI-Core score and ranking file.
4. BENI-Expanded score and ranking file.
5. Final feature/data dictionary.

## Source Workbook

Expected source location:

```text
../data/raw/Capstone_Stage_ML1_Model_Ready_Dataset.xlsx
```

The workbook already contains the completed BENI variables. This notebook does not fabricate scores; it exports and documents the final BENI fields used in the report.


````python
# Import required libraries.
from pathlib import Path
import pandas as pd
import numpy as np

# Configure pandas display.
pd.set_option("display.max_columns", 100)
pd.set_option("display.width", 120)

# Define project paths.
PROJECT_ROOT = Path("..")
RAW_DIR = PROJECT_ROOT / "data" / "raw"
PROCESSED_DIR = PROJECT_ROOT / "data" / "processed"

# Ensure output directory exists.
PROCESSED_DIR.mkdir(parents=True, exist_ok=True)

# Define source workbook path.
WORKBOOK_PATH = RAW_DIR / "Capstone_Stage_ML1_Model_Ready_Dataset.xlsx"

WORKBOOK_PATH
````

## 1. Load BENI Source Sheets

The BENI-Core and BENI-Expanded fields are loaded from the project workbook.

````python
# Load source sheets.
core_df = pd.read_excel(WORKBOOK_PATH, sheet_name="Model_Ready_Core_StateYear")
expanded_df = pd.read_excel(WORKBOOK_PATH, sheet_name="Model_Ready_Expanded_2022")
feature_dictionary_df = pd.read_excel(WORKBOOK_PATH, sheet_name="Feature_Dictionary")
target_definitions_df = pd.read_excel(WORKBOOK_PATH, sheet_name="Target_Definitions")

# Confirm shapes.
print("Core sheet shape:", core_df.shape)
print("Expanded sheet shape:", expanded_df.shape)
print("Feature dictionary shape:", feature_dictionary_df.shape)
print("Target definitions shape:", target_definitions_df.shape)
````

## 2. BENI-Core Dataset

BENI-Core is based on UBEC-derived education infrastructure pressure indicators.

Core indicators include:

```text
pupil_teacher_ratio
pupil_classroom_ratio
unqualified_teacher_share
bad_classroom_share
enrolment_per_school
```


````python
# Export the final BENI-Core model-ready dataset.
beni_core_model_ready = core_df.copy()

# Add repository metadata.
beni_core_model_ready["processed_source_file"] = "Capstone_Stage_ML1_Model_Ready_Dataset.xlsx"
beni_core_model_ready["processed_source_sheet"] = "Model_Ready_Core_StateYear"
beni_core_model_ready["processed_dataset_name"] = "beni_core_model_ready"
beni_core_model_ready["processed_status"] = "final_model_ready"

# Save output.
beni_core_model_ready.to_csv(PROCESSED_DIR / "beni_core_model_ready.csv", index=False)

# Display summary.
print("BENI-Core rows:", beni_core_model_ready.shape[0])
print("BENI-Core columns:", beni_core_model_ready.shape[1])
print("BENI-Core states:", beni_core_model_ready["state"].nunique())
print("BENI-Core years:", sorted(beni_core_model_ready["year"].unique().tolist()))
print("Missing values:", int(beni_core_model_ready.isna().sum().sum()))

beni_core_model_ready.head()
````

## 3. BENI-Expanded Dataset

BENI-Expanded extends BENI-Core by adding the NBS-derived socioeconomic need proxy.

Expanded indicators include the five core education indicators plus:

```text
nbs_socioeconomic_need_proxy
```


````python
# Export the final BENI-Expanded 2022 model-ready dataset.
beni_expanded_2022_model_ready = expanded_df.copy()

# Add repository metadata.
beni_expanded_2022_model_ready["processed_source_file"] = "Capstone_Stage_ML1_Model_Ready_Dataset.xlsx"
beni_expanded_2022_model_ready["processed_source_sheet"] = "Model_Ready_Expanded_2022"
beni_expanded_2022_model_ready["processed_dataset_name"] = "beni_expanded_2022_model_ready"
beni_expanded_2022_model_ready["processed_status"] = "final_model_ready"

# Save output.
beni_expanded_2022_model_ready.to_csv(PROCESSED_DIR / "beni_expanded_2022_model_ready.csv", index=False)

# Display summary.
print("BENI-Expanded rows:", beni_expanded_2022_model_ready.shape[0])
print("BENI-Expanded columns:", beni_expanded_2022_model_ready.shape[1])
print("BENI-Expanded states:", beni_expanded_2022_model_ready["state"].nunique())
print("BENI-Expanded years:", sorted(beni_expanded_2022_model_ready["year"].unique().tolist()))
print("Missing values:", int(beni_expanded_2022_model_ready.isna().sum().sum()))

beni_expanded_2022_model_ready.head()
````

## 4. Export BENI-Core Scores

This file contains the final BENI-Core scores, ranks, and priority bands for the 148 state-year records.

````python
# Select BENI-Core score and ranking fields.
beni_core_scores = core_df[
    [
        "state",
        "year",
        "target_beni_core_score",
        "beni_rank_within_year",
        "priority_band_within_year",
        "split_group",
        "model_ready_flag",
        "source_note",
    ]
].copy()

# Add repository metadata.
beni_core_scores["source_file"] = "Capstone_Stage_ML1_Model_Ready_Dataset.xlsx"
beni_core_scores["source_sheet"] = "Model_Ready_Core_StateYear"
beni_core_scores["processed_dataset_name"] = "beni_core_scores"
beni_core_scores["processed_status"] = "final_processed_score_file"

# Save output.
beni_core_scores.to_csv(PROCESSED_DIR / "beni_core_scores.csv", index=False)

# Display preview.
beni_core_scores.head()
````

## 5. Export BENI-Expanded Scores

This file contains the final BENI-Expanded score, rank, socioeconomic proxy, population context, and fiscal-context variables for the 37 states in 2022.

````python
# Select BENI-Expanded score and context fields.
beni_expanded_scores = expanded_df[
    [
        "state",
        "year",
        "target_beni_expanded_score",
        "beni_expanded_rank",
        "target_beni_core_score",
        "beni_core_rank",
        "nbs_socioeconomic_need_proxy",
        "school_age_basic_population",
        "dmo_domestic_debt_stock_naira",
        "fiscal_pressure_band",
        "split_group",
        "model_ready_flag",
        "source_note",
    ]
].copy()

# Add repository metadata.
beni_expanded_scores["source_file"] = "Capstone_Stage_ML1_Model_Ready_Dataset.xlsx"
beni_expanded_scores["source_sheet"] = "Model_Ready_Expanded_2022"
beni_expanded_scores["processed_dataset_name"] = "beni_expanded_scores"
beni_expanded_scores["processed_status"] = "final_processed_score_file"

# Save output.
beni_expanded_scores.to_csv(PROCESSED_DIR / "beni_expanded_scores.csv", index=False)

# Display top 10 states by BENI-Expanded rank.
beni_expanded_scores.sort_values("beni_expanded_rank").head(10)
````

## 6. Top 10 BENI-Expanded Priority States

This table is used directly in the Preliminary Results section of the report.

````python
# Extract top 10 BENI-Expanded priority states.
top_10_beni_expanded = beni_expanded_scores.sort_values("beni_expanded_rank").head(10)

# Display compact priority table.
top_10_beni_expanded[
    [
        "beni_expanded_rank",
        "state",
        "target_beni_expanded_score",
        "target_beni_core_score",
        "nbs_socioeconomic_need_proxy",
        "fiscal_pressure_band",
    ]
]
````

## 7. Export Final Dataset Dictionary

This file supports the data dictionary requirement in the report and appendices.

````python
# Add metadata to feature dictionary.
final_dataset_dictionary = feature_dictionary_df.copy()
final_dataset_dictionary["source_file"] = "Capstone_Stage_ML1_Model_Ready_Dataset.xlsx"
final_dataset_dictionary["source_sheet"] = "Feature_Dictionary"
final_dataset_dictionary["processed_dataset_name"] = "final_dataset_dictionary"
final_dataset_dictionary["processed_status"] = "final_processed_dictionary"

# Save output.
final_dataset_dictionary.to_csv(PROCESSED_DIR / "final_dataset_dictionary.csv", index=False)

# Display dictionary.
final_dataset_dictionary
````

## 8. BENI Interpretation Notes

The BENI scores are composite index values. They are used for ranking relative infrastructure need across states.

Important interpretation rules:

1. A higher BENI score indicates greater relative basic education infrastructure need.
2. BENI-Core reflects education infrastructure pressure only.
3. BENI-Expanded reflects education infrastructure pressure plus socioeconomic need.
4. BENI is a decision-support index, not a final government allocation rule.
5. Machine-learning validation should be interpreted as reproducibility and ranking stability, not causal prediction.
