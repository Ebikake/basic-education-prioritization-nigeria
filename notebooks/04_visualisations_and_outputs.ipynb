# 04 Visualisations and Outputs

This notebook generates the report-ready tables, figures, and screenshot-preview outputs for the capstone project:

**A Data-Driven Framework for Prioritizing Public Capital Allocation to Basic Education Infrastructure across Nigerian States Using Education Need Indicators**

## Purpose

This notebook produces:

1. Core model RMSE comparison chart.
2. Expanded model mean RMSE comparison chart.
3. Actual versus predicted BENI-Expanded scatter plot.
4. Top 10 ML-predicted priority states bar chart.
5. BENI-Expanded coefficient-importance chart.
6. Dataset-preview tables for Appendix B.
7. Figure and output log for repository documentation.

## Expected Input Files

This notebook expects that `03_model_training_and_validation.ipynb` has already produced the required files in `outputs/tables/`.


````python
# Import required libraries.
from pathlib import Path
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Configure pandas display.
pd.set_option("display.max_columns", 100)
pd.set_option("display.width", 120)

# Define project paths.
PROJECT_ROOT = Path("..")
PROCESSED_DIR = PROJECT_ROOT / "data" / "processed"
TABLES_DIR = PROJECT_ROOT / "outputs" / "tables"
FIGURES_DIR = PROJECT_ROOT / "outputs" / "figures"
DOCS_DIR = PROJECT_ROOT / "docs"

# Create output directories.
FIGURES_DIR.mkdir(parents=True, exist_ok=True)
TABLES_DIR.mkdir(parents=True, exist_ok=True)
DOCS_DIR.mkdir(parents=True, exist_ok=True)
````

## 1. Load Input Tables and Datasets

````python
# Load model output tables.
core_model_performance = pd.read_csv(TABLES_DIR / "core_model_performance.csv")
expanded_model_performance = pd.read_csv(TABLES_DIR / "expanded_model_performance.csv")
expanded_ranking_comparison = pd.read_csv(TABLES_DIR / "expanded_beni_ranking_comparison.csv")
top_10_priority_states = pd.read_csv(TABLES_DIR / "top_10_priority_states.csv")
coefficient_importance = pd.read_csv(TABLES_DIR / "coefficient_importance_beni_expanded.csv")

# Load processed datasets for preview outputs.
integrated_state_year_dataset = pd.read_csv(PROCESSED_DIR / "integrated_state_year_dataset.csv")
beni_core_model_ready = pd.read_csv(PROCESSED_DIR / "beni_core_model_ready.csv")
beni_expanded_2022_model_ready = pd.read_csv(PROCESSED_DIR / "beni_expanded_2022_model_ready.csv")

# Confirm shapes.
print("Core model performance:", core_model_performance.shape)
print("Expanded model performance:", expanded_model_performance.shape)
print("Expanded ranking comparison:", expanded_ranking_comparison.shape)
print("Top 10 priority states:", top_10_priority_states.shape)
print("Coefficient importance:", coefficient_importance.shape)
````

## 2. Figure C1: Core Model RMSE Comparison

This figure supports Appendix C1 and Section 6.5.


````python
# Create Core Model RMSE comparison chart.
plt.figure(figsize=(9, 5))
plt.bar(core_model_performance["Model"], core_model_performance["RMSE"])
plt.title("Core Model RMSE Comparison on 2022 Holdout Test Set")
plt.xlabel("Model")
plt.ylabel("RMSE")
plt.xticks(rotation=30, ha="right")
plt.tight_layout()

# Save figure.
figure_path = FIGURES_DIR / "figure_c1_core_model_rmse.png"
plt.savefig(figure_path, dpi=300, bbox_inches="tight")
plt.show()

print("Saved:", figure_path)
````

## 3. Figure C2: Expanded Model Mean RMSE Comparison

This figure supports Appendix C2 and Section 6.5.


````python
# Create Expanded Model Mean RMSE comparison chart.
plt.figure(figsize=(9, 5))
plt.bar(expanded_model_performance["Model"], expanded_model_performance["Mean RMSE"])
plt.title("Expanded Model Mean RMSE from 5-Fold Cross-Validation")
plt.xlabel("Model")
plt.ylabel("Mean RMSE")
plt.xticks(rotation=30, ha="right")
plt.tight_layout()

# Save figure.
figure_path = FIGURES_DIR / "figure_c2_expanded_model_rmse.png"
plt.savefig(figure_path, dpi=300, bbox_inches="tight")
plt.show()

print("Saved:", figure_path)
````

## 4. Figure C3: Actual versus Predicted BENI-Expanded Score

This figure supports Appendix C3 and validates BENI-Expanded reproducibility.


````python
# Identify actual and predicted score columns.
actual_col = "target_beni_expanded_score"
pred_col = "ml_predicted_beni_expanded_score"

# Create scatter plot.
plt.figure(figsize=(6, 6))
plt.scatter(expanded_ranking_comparison[actual_col], expanded_ranking_comparison[pred_col])

# Add 45-degree reference line.
min_value = min(expanded_ranking_comparison[actual_col].min(), expanded_ranking_comparison[pred_col].min())
max_value = max(expanded_ranking_comparison[actual_col].max(), expanded_ranking_comparison[pred_col].max())
plt.plot([min_value, max_value], [min_value, max_value])

plt.title("Actual versus Predicted BENI-Expanded Score")
plt.xlabel("Actual BENI-Expanded Score")
plt.ylabel("Predicted BENI-Expanded Score")
plt.tight_layout()

# Save figure.
figure_path = FIGURES_DIR / "figure_c3_actual_vs_predicted_beni_expanded.png"
plt.savefig(figure_path, dpi=300, bbox_inches="tight")
plt.show()

print("Saved:", figure_path)
````

## 5. Figure C4: Top 10 ML-Predicted Priority States

This figure supports Appendix C4 and the Preliminary Results section.


````python
# Prepare top 10 states by predicted rank.
top_10_predicted = expanded_ranking_comparison.sort_values(
    "ml_predicted_beni_expanded_rank"
).head(10)

# Create bar chart.
plt.figure(figsize=(10, 5))
plt.bar(top_10_predicted["state"], top_10_predicted["ml_predicted_beni_expanded_score"])
plt.title("Top 10 ML-Predicted Priority States by BENI-Expanded Score")
plt.xlabel("State")
plt.ylabel("Predicted BENI-Expanded Score")
plt.xticks(rotation=30, ha="right")
plt.tight_layout()

# Save figure.
figure_path = FIGURES_DIR / "figure_c4_top_10_predicted_priority_states.png"
plt.savefig(figure_path, dpi=300, bbox_inches="tight")
plt.show()

print("Saved:", figure_path)
````

## 6. Figure C5: Linear Regression Coefficient Importance for BENI-Expanded

This figure supports Appendix C5 and helps explain the best-performing expanded model.


````python
# Sort coefficient importance by absolute coefficient.
coef_plot_df = coefficient_importance.sort_values("absolute_coefficient", ascending=True)

# Create horizontal bar chart.
plt.figure(figsize=(9, 5))
plt.barh(coef_plot_df["feature"], coef_plot_df["absolute_coefficient"])
plt.title("Linear Regression Coefficient Importance for BENI-Expanded")
plt.xlabel("Absolute Coefficient")
plt.ylabel("Feature")
plt.tight_layout()

# Save figure.
figure_path = FIGURES_DIR / "figure_c5_coefficient_importance_beni_expanded.png"
plt.savefig(figure_path, dpi=300, bbox_inches="tight")
plt.show()

print("Saved:", figure_path)
````

## 7. Appendix B Dataset Preview Tables

These exported preview tables support Appendix B screenshots and repository evidence.


````python
# Create processed dataset preview for Appendix B.
processed_dataset_preview = integrated_state_year_dataset.head(10).copy()
processed_dataset_preview.to_csv(TABLES_DIR / "appendix_b_processed_dataset_preview.csv", index=False)

# Create BENI-Expanded model-ready preview for Appendix B.
beni_expanded_preview = beni_expanded_2022_model_ready.head(10).copy()
beni_expanded_preview.to_csv(TABLES_DIR / "appendix_b_beni_expanded_model_ready_preview.csv", index=False)

# Create dataset components summary.
dataset_components_summary = pd.DataFrame([
    {
        "dataset": "integrated_state_year_dataset.csv",
        "rows": integrated_state_year_dataset.shape[0],
        "columns": integrated_state_year_dataset.shape[1],
        "purpose": "Final integrated state-year dataset"
    },
    {
        "dataset": "beni_core_model_ready.csv",
        "rows": beni_core_model_ready.shape[0],
        "columns": beni_core_model_ready.shape[1],
        "purpose": "BENI-Core model-ready dataset"
    },
    {
        "dataset": "beni_expanded_2022_model_ready.csv",
        "rows": beni_expanded_2022_model_ready.shape[0],
        "columns": beni_expanded_2022_model_ready.shape[1],
        "purpose": "BENI-Expanded 2022 model-ready dataset"
    },
])
dataset_components_summary.to_csv(TABLES_DIR / "dataset_components_summary.csv", index=False)

# Display summary.
dataset_components_summary
````

## 8. Cleaning and Missingness Summary

This output supports the Analysis section and Appendix evidence.


````python
# Create cleaning/missingness summary for key processed datasets.
cleaning_summary = pd.DataFrame([
    {
        "dataset": "beni_core_model_ready.csv",
        "rows": beni_core_model_ready.shape[0],
        "columns": beni_core_model_ready.shape[1],
        "missing_values": int(beni_core_model_ready.isna().sum().sum()),
        "states": beni_core_model_ready["state"].nunique(),
        "years": ", ".join(map(str, sorted(beni_core_model_ready["year"].unique().tolist()))),
        "status": "model-ready"
    },
    {
        "dataset": "beni_expanded_2022_model_ready.csv",
        "rows": beni_expanded_2022_model_ready.shape[0],
        "columns": beni_expanded_2022_model_ready.shape[1],
        "missing_values": int(beni_expanded_2022_model_ready.isna().sum().sum()),
        "states": beni_expanded_2022_model_ready["state"].nunique(),
        "years": ", ".join(map(str, sorted(beni_expanded_2022_model_ready["year"].unique().tolist()))),
        "status": "model-ready"
    },
    {
        "dataset": "integrated_state_year_dataset.csv",
        "rows": integrated_state_year_dataset.shape[0],
        "columns": integrated_state_year_dataset.shape[1],
        "missing_values": int(integrated_state_year_dataset.isna().sum().sum()),
        "states": integrated_state_year_dataset["state"].nunique(),
        "years": ", ".join(map(str, sorted(integrated_state_year_dataset["year"].unique().tolist()))),
        "status": "integrated processed dataset"
    },
])

# Save cleaning summary.
cleaning_summary.to_csv(TABLES_DIR / "cleaning_summary.csv", index=False)

# Display summary.
cleaning_summary
````
